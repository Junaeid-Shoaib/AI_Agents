# Creating Tensor

```python
#using empty
x = torch.empty(2,3)
print(x)
```

```python
#check type
type(x)
```

```python
#using zero
x = torch.zeros(2,3)
print(x)
```

```python
#using one
x = torch.ones(2,3)
print(x)
```

```python
#using rand
x = torch.rand(2,3)
print(x)
```

```python
#using manual seed for reproducibilty
torch.manual_seed(42)
x = torch.rand(2,3)
print(x)
```

```python
torch.manual_seed(42)
x = torch.rand(2,3)
print(x)
```

```python
#using tensor
x = torch.tensor([[1,2,3],[4,5,6]])
print(x)
```

```python
#using arange
x = torch.arange(0,32,2).reshape(4,4)
print(x)
```

```python
#using linspace
x = torch.linspace(0,32,16).reshape(4,4)
print(x)
```

```python
#using full
x = torch.full((2,3),4)
print(x)
```

```python
#using eye
x = torch.eye(3)
print(x)
```

```python
import time
import torch

size = 10000
m1 = torch.rand(size,size)
m2 = torch.rand(size,size)

start = time.time()
mr = torch.matmul(m1,m2)
end = time.time()
cpu = end-start
print(f'CPU: {cpu}')

m1_gpu = m1.to('cuda')
m2_gpu = m2.to('cuda')
start1 = time.time()
mr_gpu = torch.matmul(m1_gpu,m2_gpu)
end1 = time.time()
gpu = end1-start1
print(f'GPU: {gpu}')

print(cpu/gpu)

```
